```python
import pandas as pd
import kagglehub
import matplotlib.pyplot as plt
from pathlib import Path

"""
======== Files : ========
EdStatsCountry.csv
EdStatsCountry-Series.csv
EdStatsData.csv
EdStatsFootNote.csv
EdStatsSeries.csv
=========================
"""

path = Path(kagglehub.dataset_download("theworldbank/education-statistics/versions/45")) / "edstats-csv-zip-32-mb-"

# Loading the dataset into a pandas dataframe
country = pd.read_csv(path / "EdStatsCountry.csv")
series = pd.read_csv(path / "EdStatsSeries.csv")
country_series = pd.read_csv(path / "EdStatsCountry-Series.csv")
footnote = pd.read_csv(path / "EdStatsFootNote.csv")
data = pd.read_csv(path / "EdStatsData.csv")
```

```python
# The last column of the each dataframe is empty.
# We can drop it to avoid any issues when merging dataframes later on.
country = country.drop(columns=["Unnamed: 31"])
series = series.drop(columns=["Unnamed: 20"])
country_series = country_series.drop(columns=["Unnamed: 3"])
footnote = footnote.drop(columns=["Unnamed: 4"])
data = data.drop(columns=["Unnamed: 69"])
```

```python
# The first 91626 lines of the data dataframe are about regions and continents.
# By using this data along with countries' data, people would be counted twice.
# The first line of data concerning countries starts at line 91627.
starting_index = 91625

data = data.iloc[starting_index:]

data = data.drop(columns=["2020", "2025", "2030", "2035", "2040", "2045", "2050", "2055", "2060", "2065", "2070", "2075", "2080", "2085", "2090", "2095", "2100"])
```

```python
# The following code displays the percentage of filled cells per year in the dataset.
# It calculates the percentage of non-null values for each year column.
# Then it plots a bar chart to visualize the data completeness over time.

year_cols = [c for c in data.columns if c.isdigit()]
pct = (data[year_cols].notna().mean() * 100).sort_index()
pct.plot.bar(figsize=(14, 6), color="#9db4c8")
plt.ylabel("Filled cells (%)")
plt.title("Percentage of filled cells per year")
plt.show()
```

We can visualize that the year 2010 contains the most data.

## Points to Watch

- The names of the key columns differ depending on the file: `Country Code` / `CountryCode` and `Indicator Code` / `Series Code` / `SeriesCode`. You must rename them before performing a link between them.
- Indicator codes sometimes differ by a space at the end of the code or by case (`'SE.SEC. DURS.LO '`, `UIS.E.1.Guk`). You must normalize them using `.str.strip().str.upper()`.
- `footnote.Year` is text. You can extract the year using `footnote[“Year”].str[2:].astype(int)`.
- 462 indicators have different names in `data` and in `series`.
