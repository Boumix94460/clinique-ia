## The five selected indicators



| Indicator | Code | Why it matters |
|---|---|---|
| Population aged 15-24 | `SP.POP.1524.TO.UN` | The population aged 15-24 is our market since we want to focus on secondary and tertiary education. |
| Population growth (% per year) | `SP.POP.GROW` | Shows whether the market is likely to grow or not. |
| Upper-secondary students (number) | `UIS.E.3` | The direct target audience: students in upper secondary education. |
| Tertiary students (number) | `SE.TER.ENRL` | The second target audience: students in tertiary education. |
| GDP per capita, PPP (current international $) | `NY.GDP.PCAP.PP.CD` | Indicates whether people can afford our products. |

```python
# Each value = mean of the 3 reference years 2010, 2013 and 2015 (missing years are ignored).
# Company goal: sell more standard-quality (affordable) products in poorer countries,
# so we keep only "Low income" and "Lower middle income" countries.
years = ["2010", "2013", "2015"]

target_groups = ["Low income", "Lower middle income"]
target_countries = country.loc[country["Income Group"].isin(target_groups), ["Country Code", "Short Name", "Income Group"]]

indicators = {
    "sp.pop.1524.to.un": "Population 15-24",
    "sp.pop.grow": "Population growth (%)",
    "uis.e.3": "Students upper secondary",
    "se.ter.enrl": "Students upper tertiary",
    "ny.gdp.pcap.pp.cd": "GDP per capita PPA",
}

selection = data[data["Country Code"].isin(target_countries["Country Code"])
                 & data["Indicator Code"].isin(indicators)].copy()
selection["mean_2010_2013_2015"] = selection[years].mean(axis=1)

table = selection.pivot_table(index="Country Code", columns="Indicator Code",
                              values="mean_2010_2013_2015")
table = table.rename(columns=indicators)[list(indicators.values())]

# Add the country name and income group to make the table readable
table = target_countries.set_index("Country Code").join(table).set_index("Short Name")


# We keep the countries that have a value for the 5 indicators
table_q5 = table.dropna().drop(columns=["Region"], errors="ignore").sort_values("Population 15-24", ascending=False)

# Big table with pandas: all the countries, readable numbers
pd.set_option("display.max_rows", None)
pd.set_option("display.width", 200)
pd.set_option("display.float_format", "{:,.1f}".format)

table_q5
```

**Method: mean of the ranks.**
1. For each of the 5 indicators, the 51 countries are ranked from 1 (best) to 51 (worst).
   For our goal every indicator is *"the higher, the better"*: more young people, faster population growth, more students, more purchasing power.
2. Each country gets the **mean of its 5 ranks**.
3. The country with the **lowest mean rank** is the most promising one.

Ranks are used instead of raw values because the indicators have very different scales (millions of people vs a growth rate in %).
With ranks, each indicator weighs the same.

```python
indicator_cols = list(indicators.values())

ranks = table_q5[indicator_cols].rank(ascending=False, method="min").astype(int)
ranks.columns = ["Rank - " + c for c in indicator_cols]

ranking = table_q5[["Income Group"]].join(ranks)
ranking["Mean rank"] = ranks.mean(axis=1)
ranking = ranking.sort_values("Mean rank")
ranking.insert(0, "Final rank", range(1, len(ranking) + 1))

print("Countries ranked:", len(ranking))
ranking
```

```python
# Real values of the 5 indicators for the top 15 countries (in ranking order, 1st at the top).
# One small chart per indicator, because the indicators do not have the same unit.
top15 = ranking.head(15).index
values = table_q5.loc[top15, indicator_cols]

def short(v):
    # 1,234,567 -> "1.2 M", 45,600 -> "46 k"
    if v >= 1e6:
        return f"{v / 1e6:.1f} M"
    if v >= 1e3:
        return f"{v / 1e3:.0f} k"
    return f"{v:.0f}"

formats = {
    "Population 15-24": short,
    "Population growth (%)": lambda v: f"{v:.1f} %",
    "Students upper secondary": short,
    "Students upper tertiary": short,
    "GDP per capita PPA": lambda v: f"{v:,.0f} $",
}
colors = ["#3b6ea5", "#4f9a62", "#c8753a", "#a5486b", "#6a5aa5"]

fig, axes = plt.subplots(1, 5, figsize=(18, 7), sharey=True)
y = range(len(top15))
for ax, col, color in zip(axes, indicator_cols, colors):
    v = values[col]
    ax.barh(y, v, color=color)
    for i, x in enumerate(v):
        ax.text(max(x, 0), i, " " + formats[col](x), va="center", fontsize=8.5)
    ax.set_title(col, fontsize=11)
    ax.set_xlim(min(0, v.min() * 3), v.max() * 1.35)   # room for the labels (and for negative values)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xticks([])
    ax.tick_params(axis="y", length=0)
    for side in ["top", "right", "bottom", "left"]:
        ax.spines[side].set_visible(False)

axes[0].set_yticks(list(y))
axes[0].set_yticklabels([f"{r}. {c}" for r, c in zip(ranking.head(15)["Final rank"], top15)])
axes[0].invert_yaxis()

fig.suptitle("Top 15 countries: value of each indicator (mean of 2010, 2013 and 2015)", fontsize=14)
plt.tight_layout()
plt.show()
```
